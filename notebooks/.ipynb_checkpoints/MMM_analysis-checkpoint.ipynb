# Bayesian Marketing Mix Model - MMM Analysis

```python
# Load data
data_path = '../data/dt_simulated_weekly.csv'
df = pd.read_csv(data_path)

print(f"Data shape: {df.shape}")
print(f"\nColumns: {list(df.columns)}")
print(f"\nFirst few rows:")
print(df.head())
```

```python
# Configuration - COLUMN NAMES MATCH CSV
config = {
    'date_col': 'DATE',
    'spend_cols': ['tv_S', 'ooh_S', 'print_S', 'facebook_S', 'search_S'],
    'revenue_col': 'revenue',
    'control_cols': ['competitor_sales_B', 'newsletter'],
    'fourier_k': 3,
    'mcmc_params': {
        'draws': 1000,
        'tune': 1000,
        'target_accept': 0.9
    }
}

print("✓ Configuration set")
print(f"  Spend columns: {config['spend_cols']}")
print(f"  Control columns: {config['control_cols']}")
print(f"  Revenue column: {config['revenue_col']}")
```

```python
# Initialize trainer
trainer = BayesianMMMTrainer(
    config,
    data_path,
    '../data/dt_prophet_holidays.csv'
)
print("✓ Trainer initialized")
```

```python
# Run full analysis
try:
    roi_results = trainer.run_full_analysis(experiment_name='mmm_notebook')
    print("\n✓ Model training complete!")
except Exception as e:
    print(f"Error during training: {e}")
    import traceback
    traceback.print_exc()
    roi_results = {}
```

```python
# Display ROI results by channel
if roi_results:
    results_df = pd.DataFrame(roi_results).T
    results_df = results_df[['unscaled_roi', 'mean_beta', 'mean_alpha']].round(3)
    results_df.columns = ['ROI', 'Beta (Effectiveness)', 'Alpha (Decay)']
    print("\nChannel ROI Summary:")
    print(results_df)
else:
    print("No results available")
```
